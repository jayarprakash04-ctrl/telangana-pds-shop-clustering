```python
shop_features = master.groupby("shopNo").agg(
    distCode=("distCode", "first"),
    distName=("distName", "first"),
    total_trans=("noOfTrans", "sum"),
    avg_monthly_trans=("noOfTrans", "mean"),
    trans_std=("noOfTrans", "std"),
    total_rice=("riceAfsc", "sum"),
    total_rice_fsc=("riceFsc", "sum"),
    total_rice_aap=("riceAap", "sum"),
    total_wheat=("wheat", "sum"),
    total_sugar=("sugar", "sum"),
    total_kerosene=("kerosene", "sum"),
    total_portability=("otherShopTransCnt", "sum"),
    totalRcs=("totalRcs", "first"),
    latitude=("latitude", "first"),
    longitude=("longitude", "first"),
).reset_index()

shop_features.head()
```

```python
eps = 1e-6

# Combine all rice sub-schemes into one total
shop_features["total_rice_all"] = (
    shop_features["total_rice"] + shop_features["total_rice_fsc"] + shop_features["total_rice_aap"]
)

# Utilization Ratio: MONTHLY rate vs entitlement (comparable across shops regardless of data span)
shop_features["utilization_ratio"] = shop_features["avg_monthly_trans"] / (shop_features["totalRcs"] + eps)

# Portability Ratio: share of transactions coming from non-local (ONORC) beneficiaries
shop_features["portability_ratio"] = shop_features["total_portability"] / (shop_features["total_trans"] + eps)

# Volatility: coefficient of variation of monthly transactions
shop_features["volatility_cv"] = shop_features["trans_std"] / (shop_features["avg_monthly_trans"] + eps)

# Rice dependency: what share of total commodity volume is rice
shop_features["commodity_total"] = (
    shop_features["total_rice_all"] + shop_features["total_wheat"] + shop_features["total_sugar"]
)
shop_features["rice_dependency"] = shop_features["total_rice_all"] / (shop_features["commodity_total"] + eps)

shop_features = shop_features.fillna(0)
shop_features[["utilization_ratio", "rice_dependency", "portability_ratio", "volatility_cv"]].describe()
```

```python
# Use a k-distance plot to help choose eps
from sklearn.neighbors import NearestNeighbors

neighbors = NearestNeighbors(n_neighbors=5).fit(X_scaled)
distances, _ = neighbors.kneighbors(X_scaled)
distances = np.sort(distances[:, -1])

plt.figure(figsize=(8, 4))
plt.plot(distances)
plt.title("K-distance plot (look for the 'elbow' - that's a good eps)")
plt.ylabel("5th nearest neighbor distance")
plt.show()
```
